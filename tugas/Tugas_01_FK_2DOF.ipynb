```python
# Jalankan sel ini dulu — berisi fungsi yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

def forward_kinematics_2dof(theta1_deg, theta2_deg, L1, L2):
    t1 = np.radians(theta1_deg)
    t2 = np.radians(theta2_deg)
    x1 = L1 * np.cos(t1)
    y1 = L1 * np.sin(t1)
    x2 = x1 + L2 * np.cos(t1 + t2)
    y2 = y1 + L2 * np.sin(t1 + t2)
    return {'base': (0,0), 'joint2': (x1,y1), 'end_effector': (x2,y2)}

print("Fungsi siap!")
```
